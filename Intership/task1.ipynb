```python
import pandas as pd
data={
    "Student":["A","B","C","D","E"],
    "Math":[78,65,90,55,82],
    "Science":[85,70,92,60,80],
    "English":[74,68,88,58,79]
}
df=pd.DataFrame(data)
print(df)
```

```python
df['Average']=df[['Math','Science','English']].mean(axis=1)
print(df)
```

```python
sub_average=df[['Math','Science','English']].mean()
print(sub_average)
```

```python
topper=df.loc[df['Average'].idxmax()]
print(topper)
```

```python
print("\nMissing Values\n",df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
plt.bar(df['Student'],df['Average'])
plt.xlabel('Student')
plt.ylabel('Average_marks')
plt.show()
```

```python
sub_average.plot(kind='bar')
plt.xlabel('Subject_Average')
plt.ylabel('Subject')
plt.show()
```

CONCLUSIONS
1. Top student: C
2. Best subject: Science
3. Overall trend: A, C, E above average; D below average
