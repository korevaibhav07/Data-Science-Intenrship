Objective: To analyze the extracted website heading data to understand the primary business focus areas of the TRUERP ERP platform.

```python
import pandas as pd
import matplotlib.pyplot as plt
import re
from collections import Counter
```

```python
df = pd.read_csv("truerp_headings.csv")
df
```

```python
df.describe(include="all")
```

Website headings represent key business modules and priorities.
Analyzing them helps identify which services and features are emphasized.

```python
heading_dist = df["Tag"].value_counts(normalize=True) * 100
heading_dist
```

```python
heading_dist.plot(kind="bar")
plt.title("Relative Importance of Website Sections")
plt.ylabel("Percentage (%)")
plt.xlabel("Heading Level")
plt.show()
```

```python
df["Text_clean"] = (
    df["Text"]
    .astype(str)
    .str.lower()
    .str.replace(r"[^a-z ]", "", regex=True)
)

df.head()
```

The most frequent keywords indicate that the website primarily focuses on ERP functionality, services, pricing, and business growth, which aligns with manufacturing ERP solutions
