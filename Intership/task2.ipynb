```python
import pandas as pd
data={
    "Student":["A","B","C","D","E"],
    "Math":[78,65,90,55,82],
    "Science":[85,70,92,60,80],
    "English":[74,68,88,58,79]
}
df=pd.DataFrame(data)
df
```

```python
df.head(3)
```

```python
df.shape
```

```python
df.dtypes
```

```python
Total = df[['Math','Science','English']].sum(axis=1)
Total
```

```python
df['avg'] = df[['Math','Science','English']].mean(axis=1)
df['avg']
```

```python
subject_avg = df[['Math', 'Science', 'English']].mean()
subject_avg
```

```python
subject_max = df[['Math', 'Science', 'English']].max()
subject_max
```

```python
df.set_index('Student', inplace=True)
```

```python
df.loc['New','Math']=None
df.isnull().sum()
```

```python
df
```

```python
df['Math'].fillna(df['Math'].mean(),inplace=True)
df['Science'].fillna(df['Science'].mean(),inplace=True)
df['English'].fillna(df['English'].mean(),inplace=True)
```

```python
df
```

```python
topper=df['avg'].idxmax()
topper
```

```python
lower=df['avg'].idxmin()
lower
```

```python
df = df.drop('New', axis=0)
```

```python
df['Result'] = df['avg'].apply(lambda x: 'Pass' if x >= 60 else 'Fail')
df
```

```python
import matplotlib.pyplot as plt
plt.bar(df.index,df['avg'])
plt.xlabel('Student')
plt.ylabel('Average_marks')
plt.show()
```

```python
subject_avg.plot(kind='bar')
plt.xlabel('Subject_Averag')
plt.ylabel('Subject')
plt.show()
```

```python
df[['Math','Science','English']].plot(kind='line', marker='o')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.title('Marks Comparison per Subject')
plt.legend(title='Subject')
plt.show()
```
